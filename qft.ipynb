# Quantum Fourier Transform

The QFT does not change how *likely* each state is. It changes the **phases**.

Every QFT circuit is the same three ingredients:

- `H` on a qubit
- controlled phase rotations `cp` from the qubits below it
- swaps at the end to reverse the qubit order

The rotations get smaller as the qubits get further apart: `pi/2`, `pi/4`, `pi/8`.

```python
# !pip install qiskit qiskit-aer matplotlib pylatexenc

from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.quantum_info import Statevector
import numpy as np

sim = AerSimulator()
```

## n = 1  —  just a Hadamard

```python
# QFT n=1

qc = QuantumCircuit(1)

qc.h(0)

display(qc.draw(output="mpl"))
print(np.round(Statevector(qc).data, 3))
```

## n = 2

```python
# QFT n=2

qc = QuantumCircuit(2)

qc.h(1)
qc.cp(np.pi/2, 0, 1)
qc.h(0)
qc.swap(0, 1)

display(qc.draw(output="mpl"))
print(np.round(Statevector(qc).data, 3))
```

## n = 3

```python
# QFT n=3

qc = QuantumCircuit(3)

qc.h(2)
qc.cp(np.pi/2, 1, 2)
qc.cp(np.pi/4, 0, 2)

qc.h(1)
qc.cp(np.pi/2, 0, 1)

qc.h(0)

qc.swap(0, 2)

display(qc.draw(output="mpl"))
print(np.round(Statevector(qc).data, 3))
```

## n = 4

```python
# QFT n=4

qc = QuantumCircuit(4)

qc.h(3)
qc.cp(np.pi/2, 2, 3)
qc.cp(np.pi/4, 1, 3)
qc.cp(np.pi/8, 0, 3)

qc.h(2)
qc.cp(np.pi/2, 1, 2)
qc.cp(np.pi/4, 0, 2)

qc.h(1)
qc.cp(np.pi/2, 0, 1)

qc.h(0)

qc.swap(0, 3)
qc.swap(1, 2)

display(qc.draw(output="mpl"))
print(np.round(Statevector(qc).data, 3))
```
